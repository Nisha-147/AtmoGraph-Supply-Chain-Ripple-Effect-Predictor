```python
from google.colab import files

print("Apne computer se DataCo CSV file select karein:")
uploaded = files.upload()

# Yeh code apne aap check kar lega ki kaun si file upload hui hai
import pandas as pd
import io

for filename in uploaded.keys():
    print(f"\nFile successfully uploaded: {filename}")
    df = pd.read_csv(io.BytesIO(uploaded[filename]), encoding="latin-1")
    break

print("\nDataset Shape (Rows, Columns):", df.shape)
display(df.head(3))
```

```python
import zipfile
import io
import os
import pandas as pd

# Uploaded dictionary se zip file ko read aur extract karna
for filename in uploaded.keys():
    if filename.endswith('.zip'):
        print(f"Extracting {filename}...")
        with zipfile.ZipFile(io.BytesIO(uploaded[filename]), 'r') as zip_ref:
            zip_ref.extractall("extracted_data")

        # Extract hue folder me se CSV file ka naam dhundhna
        extracted_files = os.listdir("extracted_data")
        csv_file = [f for f in extracted_files if f.endswith('.csv')][0]
        csv_path = os.path.join("extracted_data", csv_file)

        # Pandas DataFrame me load karna
        df = pd.read_csv(csv_path, encoding="latin-1")
        print(f"\nSuccessfully loaded CSV file: {csv_file}")
        break

print("\nDataset Shape (Rows, Columns):", df.shape)
display(df.head(3))
```
